# Separación de un dataset por tiempo y espacio (Sicing y Dicing)

Xarray crea indices de coordenadas etiquetadas para datos compatibles con CF. Esto hace la selección de subconjuntos de los datos en tiempo y espacio sencilla.

Este tutorial cubre alguno de los patrones usados comúnes para dividir datos.

Cargamos la librería y el datase como el notebook anterior.

```python
import xarray
```

```python
ds = xarray.open_dataset('jro20250516_132000.nc')
```

Selecciona solo las variables de datos ``ne`` y guada un referencia suya en otra variable de python

```python
ne = ds.ne
```

```python
ne
```

```python
ne.shape
```

```python
ne[0,:]
```

Seleccionando solo el primer índice de tiempo, se ha creado un DataArray sin dimendion de tiempo, pero el tiempo es aún una coordenada no asociada con algnua variable, como indica por tener más ``*`` junto a ella. Ahora sólo tiene un único valor: el valor de el primer índice de tiempo. La selección de índice anterior es equivalente a usar ``isel`` de la siguiente manera:

```python
ne.isel(timestamps=0)
```

```python
ne.isel(timestamps=slice(20,None))
```

```python
ne.isel(timestamps=0).sel(gdalt=slice(60,120))
```

En este caso es conveniente usar ``isel`` para seleccionar el tiempo, en lugar de especificar un día. Pero también es posible especificar el día usando ``sel``

Es tambien posible usar ``slice`` para la dimension ``time``. Para seleccionar Marzo a Noviembre de 1871

```python
ne.sel(timestamps=slice(1.7156175e+09, 1.715624e+09), gdalt=slice(60, 120))
```

El operador ``slice`` selecciona valores entre un valor superior e inferior. Si un valor de coordenada sola es requerida cuando usamos ``sel`` eso debe ya sea corresponder a un valor exacto en el arreglo de coordendas, o el argumento ``method`` especificado para decir a xarray como elegir un valor. Por ejemplo, para seleccionar un solo valor en la celda más cercana a los valores que deseamos.

```python
ne.sel(timestamps = 100, method='nearest')
```

Así el valor mas cercano en los datos estaba en ``timestamps = 105.0``, o lo que es lo mismo el que tiene un ``timestamps = 1.7474016e+09``
